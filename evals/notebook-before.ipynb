## Uncertainty

It is important to stress that the result is not definitive. On the reused diagnostic set, the F1 difference (A minus B) was 0.026, with a 95% interval of [-0.010, 0.061] from 2,000 paired bootstrap samples. Resampling held the fitted models and class sample counts fixed and excluded retraining and model-selection variation. The same training OOF labels selected the settings and scored the selected settings, creating selection optimism.

````python
brier_scores = {"baseline": 0.052, "weighting": 0.087}
for model, score in brier_scores.items():
    print(f"{model}: {score:.3f}")
````
